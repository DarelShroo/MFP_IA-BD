```python
%%writefile flights.pig
-- Generamos el fichero pig en este caso flights.pig

/* 
Cargamos la librería PiggyBank (un conjunto de funciones extra para Apache Pig).
Esta librería incluye CSVExcelStorage, que nos permite leer archivos CSV
correctamente (teniendo en cuenta comas, comillas, etc.).
*/
REGISTER piggybank.jar

/*
Leemos el archivo airports.csv desde el sistema de ficheros.

Para ello usamos CSVExcelStorage, que es el loader encargado de interpretar
correctamente archivos CSV (separados por comas).

Indicamos:
- ',' como separador de columnas.
- Que la primera fila del archivo es la cabecera y debe ignorarse.
*/
AIRPORTS = LOAD '$airports_file' USING
       org.apache.pig.piggybank.storage.CSVExcelStorage(',', 'NO_MULTILINE', 'UNIX', 'SKIP_INPUT_HEADER')
       AS (airportid:chararray, city:chararray, state:chararray, airportname:chararray);

/*
Cargamos el archivo flights.csv en Pig y lo guardamos en la variable FLIGHTS.

El archivo es un CSV, por lo que usamos CSVExcelStorage para leerlo correctamente
    (separación por comas, formato correcto de los campos, etc.).

Configuración del loader:
    - ',' → separador de columnas.
    - NO_MULTILINE → cada fila del archivo ocupa una sola línea.
    - UNIX → formato de salto de línea tipo Linux.
    - SKIP_INPUT_HEADER → se ignora la primera fila del archivo (cabecera).

Finalmente, definimos el nombre y el tipo de dato de cada columna
para que Pig sepa cómo interpretar los valores.
*/

FLIGHTS = LOAD '$flights_file' USING
       org.apache.pig.piggybank.storage.CSVExcelStorage(',', 'NO_MULTILINE', 'UNIX', 'SKIP_INPUT_HEADER')
       AS (dayofmonth:int, dayofweek:int, carrier:chararray, 
               depairportid:chararray, arrairportid:chararray, depdelay:int, arrdelay:int);

      
-- Nos quedamos con 10 aeropuertos
AIRPORTS_10 = LIMIT AIRPORTS 10;

-- Mostramos 10 aeropuertos
DUMP AIRPORTS_10;

-- Hacemos lo mismo con los vuelos
FLIGHTS_10 = LIMIT FLIGHTS 10;
DUMP FLIGHTS_10;
```

## 2.- Con una consulta de HQL muestra: La cinco compañías que más vuelos retrasados tienen.

* El campo `carrier` contiene la compañía aérea.
* Vamos a considerar que un vuelo llega con retraso cuando el vuelo llega más de 15 minutos tarde (campo `arrdelay` > 15).

Se espera el siguiente resultado:

![solución 2](./img/2.png)

```python
! beeline -u "jdbc:hive2://localhost:10000/bda03" -e "\
SELECT carrier, COUNT(arrdelay) AS delayed_flights  \
FROM flights \
WHERE arrdelay > 15 \
GROUP BY carrier \
ORDER BY delayed_flights DESC \
LIMIT 5;"
```

## 3.- Con una consulta de HQL muestra: Las 5 compañías que mejor recuperación de tiempo en vuelo tienen.

* Se considera que se ha recuperado el tiempo de un vuelo cuando habiendo salido con retraso (`depdelay` > 15), llega sin retraso (`arraydelay` <= 15).
* Se trata de que muestres las 5 compañías que han recuperado el tiempo en un mayor porcentaje de vuelos que salieron retrasados.

El resultado esperado es el siguiente:

![solución 3](./img/3.png)

```python
!beeline -u "jdbc:hive2://localhost:10000/bda03" -e "\
SELECT \
    carrier, \
    CAST(COUNT(CASE WHEN arrdelay <= 15 THEN 1 END) AS DOUBLE) / COUNT(*) AS percent_recovered \
FROM flights \
WHERE depdelay > 15 \
GROUP BY carrier \
ORDER BY percent_recovered DESC \
LIMIT 5"
```

## 4.- Resuelve el ejercicio 2 con Pig Latin

El resultado esperado es:

![solución 4](./img/4.png)

```python
%%writefile flights.pig

REGISTER piggybank.jar

FLIGHTS = LOAD '$flights_file' USING
       org.apache.pig.piggybank.storage.CSVExcelStorage(',', 'NO_MULTILINE', 'UNIX', 'SKIP_INPUT_HEADER')
       AS (dayofmonth:int, dayofweek:int, carrier:chararray, 
           depairportid:chararray, arrairportid:chararray, depdelay:int, arrdelay:int);

DELAYED_FLIGHTS = FILTER FLIGHTS BY arrdelay > 15;

GROUPED_CARRIER = GROUP DELAYED_FLIGHTS BY carrier;

COUNTER_ARR_DELAY = FOREACH GROUPED_CARRIER GENERATE
    group AS carrier,
    COUNT(DELAYED_FLIGHTS) AS delayed_flights;

ORDERED = ORDER COUNTER_ARR_DELAY BY delayed_flights DESC;

FLIGHTS_5 = LIMIT ORDERED 5;

DUMP FLIGHTS_5;
```

```python
! pig -x local -f flights.pig -param flights_file='flights.csv' -param output_dir='pig/output/flights'
```

## 5.- Resuelve el ejercicio 3 con Pig Latin

Se espera el siguiente resultado:

![solución 5](./img/5.png)

```python
%%writefile flights.pig

REGISTER piggybank.jar

FLIGHTS = LOAD '$flights_file' USING
       org.apache.pig.piggybank.storage.CSVExcelStorage(',', 'NO_MULTILINE', 'UNIX', 'SKIP_INPUT_HEADER')
       AS (dayofmonth:int, dayofweek:int, carrier:chararray, 
               depairportid:chararray, arrairportid:chararray, depdelay:int, arrdelay:int);
        
DELAYED_FLIGHTS = FILTER FLIGHTS BY depdelay > 15;

RECOVERED_FLIGHTS = FOREACH DELAYED_FLIGHTS GENERATE
    carrier,
    (arrdelay <= 15 ? 1 : 0) AS recovered;

GROUPED = GROUP RECOVERED_FLIGHTS BY carrier;

STATS = FOREACH GROUPED GENERATE
    group AS carrier,
    (double)SUM(RECOVERED_FLIGHTS.recovered) / COUNT(RECOVERED_FLIGHTS) AS percent_recovered;

ORDERED = ORDER STATS BY percent_recovered DESC;

FLIGHTS_5 = LIMIT ORDERED 5;

DUMP FLIGHTS_5;
```
