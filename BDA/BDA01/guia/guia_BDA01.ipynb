## Herramientas que vamos a utilizar

Instalar una infraestructura Hadoop es complicado y requiere de una infraestructura de red y de equipos muy costosa. Hay proveedores que ofrecen soluciones `cloud` para Hadoop. En la actualidad ninguno de ellos permite una prueba gratuita. Por ello vamos a hacer una instalación de `Hadoop` simplificada y virtualizada.

En lugar de utilizar un sistema de virtualización pesada como VirtualBox, vamos a hacer una virtualización ligera basada en contenedores con Docker. La virtualización con Docker permite que los contenedores compartan recursos con el SO anfitrión, con lo que es mucho más rápida y requiere de muy pocos recursos. Docker funcionará mejor y más rápido en Linux pero también se puede instalar en Windows. Para poder ejecutar el entorno, vamos a necesitar un equipo relativamente moderno con un procesador i5 o superior (no tiene porqué ser de Intel) y una memoria RAM de un mínimo de 8GB aunque funcionará mejor cuanta más memoria tenga nuestro equipo.

Para algunas tareas vamos a tener que ejecutar instrucciones de línea de comandos de la Bash o de la PowerShell de Windows. En el aula virtual se suministran enlaces a documentación de la Bash libremente accesibles.

La mayoría de los componentes del ecosistema Hadoop se programan de forma nativa en Java (algunos, como Apache Spark, en Scala, que esta basado en Java aunque es menos verboso y usa el paradigma funcional). Aunque en un entorno profesional el uso de Java con Hadoop va a ser imprescindible, para probar y aprender Hadoop puede resultar menos conveniente. Java es un lenguaje muy verboso y fuertemente tipado y requiere conocer la jerarquía de clases, métodos y tipos de las distintas herramientas y además hay que compilar, empaquetar y desplegar nuestros programas. Por ello se ha decidido utilizar Python en lugar de Java con Hadoop ya que, al ser un lenguaje más sencillo, con tipado dinámico e interpretado, va a hacer que el aprendizaje de Hadoop sea más rápido. En el aula virtual se suministran enlaces a documentación del lenguaje Python libremente accesibles.

Por último, para ejecutar Python dentro de Hadoop se usarán notebooks de Jupyter con lo que no será necesario instalar Python en el equipo anfitrión. Más adelante se explica qué es Jupyter.

Para saber la versión de `docker` instalada en un libro de Jupyter ejecutado en el anfitrión:

En `docker` las imágenes virtualizadas se llaman contenedores. Podemos crear nuestras propias imágenes escribiendo un fichero `Dockerfile` y compilándolo. No se explica como hacer esto en esta guía ya que no vamos a necesitar crear nuevas imágenes. Y para crear un contenedor de manera interactiva:

![docker run](./img/docker-run.png)

Observa que el `prompt` ha cambiado y ahora muestra el símbolo "#" del `superusuario`.
Ahora estás "dentro" del contenedor creado y cualquier comando que ejecutes se hará en el mismo. Por ejemplo, podemos saber el nombre de la máquina asignado al crear el contenedor. 

![docker hostname](./img/docker-hostname.png)

Reiteramos: el terminal que hemos abierto está ocupado por el contenedor de `docker`. Cualquier comando que ejecutamos se hace en el contenedor. Para seguir practicando comandos de `docker` necesitamos abrir un segundo terminal para que nuevamente sea el anfitrión el que reciba los comandos.

Podemos saber qué contenedores se están ejecutando:

````python
! docker ps
````

Observa que el contenedor que hemos creado tiene un `CONTAINER_ID` único y también un nombre asignado por `docker` aleatoriamente. Cuando usemos comandos de `docker` podemos usar tanto el `id` como el `name`.

Por ejemplo, podemos parar un contenedor con (el nombre será el que se haya asignado al contenedor):

````python
! docker stop modest_swirles
````

Ahora ya no habrá contenedores en ejecución:

````python
! docker ps
````

Con la opción `-a` podemos ver todos los contenedores, incluso los que estén parados.

````python
! docker ps -a
````

Podemos reiniciar un contenedor con: 

````python
! docker start modest_swirles
````

Observamos que el contenedor `modest_swirles` vuelve a estar en ejecución:

````python
! docker ps
````

Sin embargo, a diferencia de lo que ocurría antes, ahora el contenedor no tiene terminal. Podemos crear un terminal para el contenedor con:

```bash
docker exec -it modest_swirles bash
```

![docker exec](./img/docker-exec.png)

Así `exec` permite ejecutar cualquier comando en un contenedor en ejecución. No confundir con `run` que lo que hace es crear un nuevo contenedor. 

Podemos eliminar un contenedor pero para ello primero hay que pararlo:

````python
! docker stop modest_swirles
! docker rm modest_swirles
````

Ahora el contenedor no estará en ejecución ni parado ya que habrá sido destruido:

````python
! docker ps -a
````

### `docker-compose`

`docker-compose` es una herramienta para "orquestación" de contenedores `docker`. Es decir, que sirve para "levantar" varios contenedores de `docker` y que compartan servicios entre sí sin necesidad de conocer los comandos de `docker`. Podríamos decir que realiza la misma tarea de [`Kubernetes`](https://kubernetes.io/es/docs/concepts/overview/what-is-kubernetes/) pero muy simplificada. Si has instalado `docker` con Docker Desktop, `docker-compose` debería estar ya instalado en tu equipo. `docker-compose` se configura mediante un fichero con extensión [`yml`](https://es.wikipedia.org/wiki/YAML). En esta guía no vamos a explicar cómo se crea un fichero de configuración de `docker-compose` sino que haremos uso de uno ya preparado para trabajar con Hadoop.

Para probar que `docker-compose` está instalado:

````python
! docker-compose --version
````

## Uso de `Jupyter Notebooks`

`Jupyter Notebooks` es una aplicación que permite ejecutar aplicaciones en Python y otros lenguajes de forma interactiva. Es decir,
que puedes escribir un fragmento de código y ejecutarlo para ver el resultado que produce. También puedes escribir texto en formato
[`markdown`](https://www.markdownguide.org/), que es una versión simplificada de `HTML` y más fácil de escribir. Por ejemplo, esto es un título "h1" en "HTML":

```HTML
<h1>Título</h1>
```

Este es el mismo título en `markdown`:

```HTML
# Título
```

Este documento usa `markdown` para escribir los textos. En cualquier caso, normalmente podemos también usar etiquetas `HTML` dentro de un documento `markdown`.

La ejecución de código interactiva y la posibilidad de mezclar código y texto, ha hecho de Jupyter una herramienta fundamental en la disciplina de la ciencia de datos.

Los documentos de Jupyter están estructurados en celdas de dos tipos: celdas de código y celdas de Markdown. Para ejecutar una celda hay que pulsar `crtl + intro` (la combinación `shift + intro` ejecuta la celda y crea una nueva vacía debajo). Las celdas de código en Jupyter se puede ejecutar en el orden que se desee (no es necesario una ejecución secuencial). Un error habitual en Jupyter es ejecutar una celda, luego modificarla y olvidar volver a ejecutarla. Con lo que el código ejecutado estará en la versión anterior.

Debes haber completado con éxito lo explicado en apartados anteriores para poder empezar la práctica. Así, partimos del supuesto de que `docker` y `docker-compose` están instalados en tu equipo.

Lo primero que debes hacer es descargar el [fichero comprimido](https://www.mecd.es/cidead/aulavirtual/pluginfile.php/319281/mod_assign/introattachment/0/Tarea%20para%20BDA01.zip?forcedownload=1) de la tarea:

![tarea](./img/tarea.png)

Una vez descargado, descomprímelo en un directorio vacío y examina sus contendidos. Se muestra el proceso en un terminal de Linux aunque lo puedes hacer perfectamente desde el entorno gráfico. Observa que ya tienes creado el directorio `notebooks` con el libro de la tarea dentro de él para que Jupyter pueda acceder directamente. También se ha incluido el fichero de configuración de `docker-compose` para Hadoop:

![tarea-list](./img/tarea-list.png)

Comprobamos que no hay ningún contenedor de `docker` en ejecución. Los podemos parar y eliminar si así lo deseamos con las siguientes instrucciones:

```bash
docker stop $(docker ps -aq)
docker rm $(docker ps -aq)
```
Comprobamos que no hay contenedores en ejecución:

![tarea-docker](./img/tarea-docker.png)

Verificamos que estamos en el directorio donde hemos descargado los ficheros de la tarea y levantamos Hadoop usando `docker-compose`:

```bash
docker-compose -f hadoop.yml up
```

Abrimos con un navegador la URL de Jupyter [http://localhost:8889](http://localhost:8889). Deberías ver el libro de la tarea:

![tarea-notebook](./img/tarea-notebook.png)

Pulsa sobre el fichero `ipynb` para abrir el libro de la tarea:

![tarea-notebook2](./img/tarea-notebook2.png)

Reemplaza todas las celdas donde ponga TODO por lo que se pide que hagas. Cuando se solicita que subas una imagen, debes situarla en el directorio `notebook/img` y enlazarla en una celda `markdown` del libro de la tarea. Cuando hayas finalizado, crea un "zip" del directorio `notebooks` y súbelo al Aula Virtual. Por ejemplo, para crear el fichero comprimido en Linux desde el anfitrión de una alumna que se llamese Begoña Sánchez Mañas:

```bash
zip -r sanchez_manas_begona_BDA01_Tarea.zip .
```

## Cuenta en `Databricks`

En la práctica de la unidad 3 me gustaría pudiéramos usar y administrar Apache Spark en un entorno de `cloud`. La compañía [`Databricks`](https://www.databricks.com/) permite probar uno de manera gratuita. Para ello requiere registrarse y esperar la aprobación de la cuenta. Dado esta puede tardar un tiempo e incluso ser denegada, vamos a hacer el registro ahora para que haya tiempo de plantear una práctica alternativa en caso necesario.

Para registrarse hay que rellenar un formulario: [https://www.databricks.com/try-databricks](https://www.databricks.com/try-databricks)

En el campo `title` poned `student`; en el campo `company` poned CIDEAD y en correo, vuestro correo del CIDEAD.
